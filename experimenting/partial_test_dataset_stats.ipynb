```python
pth = "/data/wear3d_preprocessed/data_test_partial_frontal_APOSE_tsoli_without_bad"
```

```python
from glob import glob
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn
```

```python
files = glob(os.path.join(pth,"*.npz"))
```

```python
len(files)
```

```python
part_percentages = []

for fl in files:
    data = np.load(fl)
    part_percentages.append(data["partiality_lm"].item())
    
```

```python
seaborn.set_theme()
```

```python
plt.figure(figsize=(15,10))
plt.hist(part_percentages,bins=10)
plt.title("Percentage of visible landmarks")
plt.xlabel("Percentage")
plt.ylabel("Number of subjects")
# plt.grid()
plt.savefig("percentage_visible_landmarks.pdf",format="pdf")
plt.show()
```
